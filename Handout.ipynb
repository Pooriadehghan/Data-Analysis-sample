### mask & where

<div dir="rtl" style="text-align: right;">
# برای انتخاب جستجویی باید یه بصورت شرط های جدا از هم یا به صورت دستی مجموعه را mask کنیم
در mask کردن هر کجا که شرط درست باشد true
مقدار را برمیگرداند
برای شروط دارای بازه یا چند شرطی از دستور where و برای نشان دادن بازه بین انها از
& | ~
استفاده میکنیم به شکلی که شرط را بصورت دوبازه نوشته و مناسبات آنرا با یکی از علامت ها مشخص میکنیم

<div>

```python
#MASK
scores = np.array([19, 10, 12, 14, 16, 11, 17, 20])

#به شکل دستی تبدیل به true ,false شده
mask = [True, False, False, False, True, False, True]

#تک شرط

mask = scores > 15
```

```python
#WHERE
# VALUES
#برای جستجو های مقداری میاوان مستقیم دستور را نوشت
scores[scores > 15]

# INDEX
#برای جستجو های اندیس های واجد شرایط از where و اصول نوشتن شرط استفاده میکنیم
# np.where(scores > 15)
np.where((16 < scores) & (scores <= 20))

#میتوان با دستور where یکسان سازی و دستوری عمل کرد
np.where(scores > 15,"Pass","Fail")



# ~  not
# &  and
# |  or

# 16 < scores < 19
filter_1 = (16 < scores)
filter_2 = (scores < 20)
scores[filter_1 &  filter_2]

#اگر مجموعه چند بعدی باشد میتوان با دستور where داده های واجد شرایط را در قالب لیت هایی از سطر و ستون بدست آوزد و با دستور zip آنها را به یک مجموعه تبدیل کرد
row, col = np.where(scores > 15)
list(zip(row, col))
```

### Mathematical Operations & Data Transformation (NumPy)

```python
#1D
X = np.array([19, 10, 12, 14, 16, 11, 17, 20])

# * / + - ** %
X / 2.1

#2D
#ضرب مجموعه های هم بعد
X1 = np.array([[19, 10, 12, 14], [16, 11, 17, 20]])
X2 = np.array([[2, 4, 5, 1], [4, 5 ,6, 7]])
X1 * X2

#dot production (ضرب ماتریسی)
#ضرب یک ماتریس n بعدی در ماتریس دیگر شرایط خاصی دارد باید ماتریس x1که دارای سطر و ستون هست میتواند در ماتریسیx2 ضرب شود که تعداد سطر آن با تعداد ستون x1 برابر باشد
X1 = np.array([[19, 10, 12, 14], [16, 11, 17, 20]])
w = np.array([[5],[1],[2],[3]])

np.dot(X1, w)  #ضرب بین دو ماتریس که شرایط ضرب دارند

X1 @ w  #پایتون بصورت برنامه نویسی شده با عمللگر @ اینکار را انجام میدهد
```

### **line plot**


```python
from matplotlib import pyplot as plt


days = ["sat", "sun", "mon", "tue", "wed", "thu", "fri"]
company_1 = [100,88,97,110,111,109,126]

company_2 = [109,90,96,104,122,113,109]
plt.plot(days, company_1, linestyle="dashed", marker="P", color="k", label="Company 1")
plt.plot(days, company_2, marker="D", color="r", label="شرکت 2")

plt.xlabel("Days")
plt.ylabel("Sell")


plt.ylim([0,150])

plt.legend()

```

### scatter


```python
plt.scatter(days, company_1, label="Company 1",s=20000,alpha=0.5, edgecolors="r")

plt.ylim([0,150])
plt.show()
```

### bar

```python
plt.bar(days, company_1, label="Company 1")

plt.ylim([0,150])
plt.show()
```

### pie

```python
plt.pie(
    company_1,
    labels=days,
    autopct="%0.2f%%",
    # rotatelabels=True,
    explode=(0.5,0,0,0,0,0,0),
)
plt.show()
```
